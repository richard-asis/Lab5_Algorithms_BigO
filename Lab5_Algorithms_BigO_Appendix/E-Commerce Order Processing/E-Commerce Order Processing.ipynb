```python
# STUDENT-SPECIFIC INPUT RULES
LAST_NAME = "ASIS"
STUDENT_ID = "TUPM-26-1217"
FAVORITE_ARTIST = "EIICHIRO ODA"

SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())
NAME_LENGTH = len(LAST_NAME)

_artist_letters = [c for c in FAVORITE_ARTIST.upper() if c.isalpha()]
ARTIST_LENGTH = len(_artist_letters)                              # letters only (spaces ignored)
ARTIST_SUM = sum(ord(c) - ord("A") + 1 for c in _artist_letters)  # A=1 ... Z=26

print("LAST_NAME      :", LAST_NAME)
print("STUDENT_ID     :", STUDENT_ID)
print("FAVORITE_ARTIST:", FAVORITE_ARTIST)
print("SEED_DIGIT     :", SEED_DIGIT)
print("ID_SUM         :", ID_SUM)
print("NAME_LENGTH    :", NAME_LENGTH)
print("ARTIST_LENGTH  :", ARTIST_LENGTH)
print("ARTIST_SUM     :", ARTIST_SUM)
```

```python
#EXERCISE 1 GENERATE ORIGINAL AND ROTATED INCIDENT RECORDS
n1 = 20 + ID_SUM
base = 1000 + ID_SUM
step = NAME_LENGTH + 1

original_records = [base + i * step for i in range(n1)]

rotation_point = SEED_DIGIT % n1
rotated_records = original_records[rotation_point:] + original_records[:rotation_point]

target_incident = original_records[(ID_SUM + ARTIST_SUM) % n1]
missing_incident = target_incident + 1

print("Original records :", original_records)
print("Rotation point  :", rotation_point)
print("Rotated records  :", rotated_records)
print("Target incident  :", target_incident)
print("Missing incident :", missing_incident)
```

```python
#EXERCISE 1 LINEAR SEARCH
def linear_search(data, target):
    comparisons = 0
    for i, value in enumerate(data):
        comparisons += 1
        if value == target:
            return i, comparisons
    return -1, comparisons
```

```python
#EXERCISE 1 ROTATED BINARY SEARCH  (O(log n))
# Every element comparison is counted: equality check, sorted-half check,
# and range check.
def rotated_binary_search(data, target):
    low, high, comparisons = 0, len(data) - 1, 0
    while low <= high:
        mid = (low + high) // 2
        comparisons += 1
        if data[mid] == target:
            return mid, comparisons

        comparisons += 1
        if data[low] <= data[mid]:                 # left half is sorted
            comparisons += 1
            if data[low] <= target < data[mid]:
                high = mid - 1
            else:
                low = mid + 1
        else:                                      # right half is sorted
            comparisons += 1
            if data[mid] < target <= data[high]:
                low = mid + 1
            else:
                high = mid - 1
    return -1, comparisons
```

```python
#EXERCISE 1 SEARCH RESULTS AND COMPARISON WITH LINEAR SEARCH
def report(label, target):
    li, lc = linear_search(rotated_records, target)
    bi, bc = rotated_binary_search(rotated_records, target)
    print(f"\n--- {label}: {target} ---")
    for name, idx, cnt in (("Linear Search", li, lc), ("Rotated Binary Search", bi, bc)):
        status = f"found at index {idx}" if idx != -1 else "NOT FOUND"
        print(f"{name:<22}: {status:<18} comparisons={cnt}")
    print("Comparison reduction:", lc - bc)
    return lc, bc

ex1_found = report("Target (present)", target_incident)
ex1_missing = report("Target (absent)", missing_incident)
```

```python
#EXERCISE 2 GENERATE THE TWO SORTED ENROLLMENT DATASETS
size_a = 10 + SEED_DIGIT
size_b = 10 + ARTIST_LENGTH

dept_a = [ID_SUM + i * (SEED_DIGIT + 1) for i in range(size_a)]          # ascending
dept_b = [ARTIST_LENGTH + i * (NAME_LENGTH + 1) for i in range(size_b)]  # ascending

print("Department A dataset:", dept_a)
print("Department B dataset:", dept_b)
```

```python
#EXERCISE 2 MERGE ALGORITHM
def merge_sorted(a, b):
    merged, i, j, comparisons = [], 0, 0, 0
    while i < len(a) and j < len(b):
        comparisons += 1
        if a[i] <= b[j]:
            merged.append(a[i])
            i += 1
        else:
            merged.append(b[j])
            j += 1
    merged.extend(a[i:])      # leftovers need no comparisons
    merged.extend(b[j:])
    return merged, comparisons

consolidated, merge_comparisons = merge_sorted(dept_a, dept_b)
```

```python
#EXERCISE 2 VERIFICATION AND OUTPUT
def is_sorted(data):
    return all(data[k] <= data[k + 1] for k in range(len(data) - 1))

print("\nConsolidated dataset:", consolidated)
print("Records before  :", len(dept_a), "+", len(dept_b), "=", len(dept_a) + len(dept_b))
print("Records after   :", len(consolidated))
print("Comparisons     :", merge_comparisons)
print("Count correct   :", len(consolidated) == len(dept_a) + len(dept_b))
print("Order correct   :", is_sorted(consolidated))
print("Matches sorted():", consolidated == sorted(dept_a + dept_b))

print("\nEXERCISE 2 COMPLEXITY (m = len(A), n = len(B))")
print("Time : O(m + n)   | Comparisons: best min(m,n), worst m+n-1")
print("Space: O(m + n)   | extra list holding the merged records")
```

```python
# SORTING ALGORITHMS
def bubble_sort(data):
    a, comparisons, swaps = data.copy(), 0, 0
    for i in range(len(a)):
        swapped = False
        for j in range(len(a) - i - 1):
            comparisons += 1
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swaps += 1
                swapped = True
        if not swapped:
            break
    return a, comparisons, swaps


def selection_sort(data):
    a, comparisons, swaps = data.copy(), 0, 0
    for i in range(len(a) - 1):
        minimum = i
        for j in range(i + 1, len(a)):
            comparisons += 1
            if a[j] < a[minimum]:
                minimum = j
        if minimum != i:
            a[i], a[minimum] = a[minimum], a[i]
            swaps += 1
    return a, comparisons, swaps


def insertion_sort(data):
    a, comparisons, shifts = data.copy(), 0, 0
    for i in range(1, len(a)):
        key, j = a[i], i - 1
        while j >= 0:
            comparisons += 1
            if a[j] <= key:
                break
            a[j + 1] = a[j]
            shifts += 1
            j -= 1
        a[j + 1] = key
    return a, comparisons, shifts
```

```python
# GENERATE ORDERS AND THE THREE INPUT CONDITIONS
import random
import time

data_size = 100 + ID_SUM * 5
orders = [(ID_SUM * 3 + i * 7) % 1000 for i in range(data_size)]
rng = random.Random(SEED_DIGIT * ID_SUM)
random_orders = orders.copy()
rng.shuffle(random_orders)

conditions = {
    "Sorted":  sorted(orders),
    "Reverse": sorted(orders, reverse=True),
    "Random":  random_orders,
}

print("Dataset Size:", data_size)
print("Generated orders (first 15):", orders[:15])
for name, values in conditions.items():
    print(f"{name:<8} input (first 10):", values[:10])
```

```python
# EXECUTION-TIME MEASUREMENT
def measure(sort_function, data, trials=5):
    times = []
    for _ in range(trials):
        start = time.perf_counter()
        result = sort_function(data)
        times.append(time.perf_counter() - start)
    return result, sum(times) / trials
```

```python
# RUN ALL ALGORITHMS ON ALL CONDITIONS + VERIFY
algorithms = [
    ("Bubble Sort", bubble_sort),
    ("Selection Sort", selection_sort),
    ("Insertion Sort", insertion_sort),
]
expected = sorted(orders)
results = []   # (condition, algorithm, comparisons, movements, time, verified)

for cond_name, cond_data in conditions.items():
    for algo_name, algo in algorithms:
        (result, comps, moves), elapsed = measure(algo, cond_data)
        results.append((cond_name, algo_name, comps, moves, elapsed, result == expected))
```

```python
# SORTING PERFORMANCE COMPARISON
print(f"\n{'Condition':<10}{'Algorithm':<18}{'Comparisons':>12}{'Movements':>12}"
      f"{'Time (s)':>14}{'Sorted?':>9}")
print("-" * 75)
for cond, name, comps, moves, elapsed, ok in results:
    print(f"{cond:<10}{name:<18}{comps:>12}{moves:>12}{elapsed:>14.8f}{str(ok):>9}")

print("\nAll results correctly sorted:", all(r[5] for r in results))
```
